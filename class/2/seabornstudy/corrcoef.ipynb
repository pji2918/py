$y_1 = x*2$


```python
import numpy as np


x = np.arange(0, 10)
y1 = x * 2
np.corrcoef(x, y1)
```

$y_2 = x*0.2$


```python
y2 = x * 0.2
np.corrcoef(x, y2)
```

$y_3 = x^3$


```python
y3 = x**3
np.corrcoef(x, y3)
```

$y_4 = RNG(84)$


```python
np.random.seed(84)
y4 = np.random.randint(0, 100, size=10)
np.corrcoef(x, y4)
```

print x and y1, 2, 3, 4.


```python
print(f"x = {x}")
print(f"y1 = {y1}")
print(f"y2 = {y2}")
print(f"y3 = {y3}")
print(f"y4 = {y4}")
```

visualize


```python
result = np.corrcoef((x, y1, y2, y3, y4))
print(result)
```

```python
from matplotlib import pyplot as plt

plt.imshow(result)
plt.colorbar()
```
