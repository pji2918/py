```python
class Tv:
    # static field
    serialnum = 0

    # instance fields
    def __init__(self, on: bool, ch: int, vol: int):
        self.on = on
        self.channel = ch
        self.volume = vol
        Tv.serialnum += 1
        self.num = Tv.serialnum

    def print(self):
        print(self.on, self.channel, self.volume, self.num)


t1 = Tv(True, 11, 10)
t1.print()
t2 = Tv(True, 9, 20)
t2.print()
```

```python
class Vector2D:
    def __init__(self, x: int, y: int):
        self.x = x
        self.y = y

    def __eq__(self, o):
        return self.x == o.x and self.y == o.y

    def __add__(self, o) -> Vector2D:
        return Vector2D(self.x + o.x, self.y + o.y)

    def __str__(self):
        return f"({self.x}, {self.y})"


v1 = Vector2D(2, 3)
v2 = Vector2D(1, 2)
print(v1 == v2)
v3 = v1 + v2
print(v3)
```

```python
class Fraction:
    def __init__(self, ja: int, mo: int):
        self.ja = ja
        self.mo = mo

    def __add__(self, other):
        if self.mo != other.mo:
            return Fraction(
                (self.ja * other.mo) + (other.ja * self.mo), self.mo * other.mo
            )
        else:
            return Fraction(self.ja + other.ja, self.mo + other.mo)

    def __mul__(self, other):
        return Fraction(self.ja * other.ja, self.mo * other.mo)

    def __str__(self):
        return f"{self.ja}/{self.mo}"


f1 = Fraction(1, 2)
f2 = Fraction(2, 3)
print(f1 + f2)
print(f1 * f2)
```

```python
class MathUtils:
    @staticmethod
    def add(a, b):
        return a + b

    @staticmethod
    def is_even(n):
        return n % 2 == 0


result = MathUtils.add(5, 3)
print(result)
```

```python
class User:
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

    @classmethod
    def from_string(cls, user_str: str):
        name, age = user_str.split(",")
        return cls(name, int(age))

    def __str__(self):
        return f"{self.name}, {self.age} year{'s' if self.age >= 2 else ''} old"


print(User.from_string("홍길동, 30"))
```

```python
class Fraction:
    def __init__(self, ja: int, mo: int):
        self.ja = ja
        self.mo = mo

    def __add__(self, other):
        if self.mo != other.mo:
            return Fraction(
                (self.ja * other.mo) + (other.ja * self.mo), self.mo * other.mo
            )
        else:
            return Fraction(self.ja + other.ja, self.mo + other.mo)

    def __mul__(self, other):
        return Fraction(self.ja * other.ja, self.mo * other.mo)

    @classmethod
    def mul2(cls, f1, f2):
        return cls(f1.ja * f2.ja, f1.mo * f2.mo)

    def __str__(self):
        return f"{self.ja}/{self.mo}"


f1 = Fraction(1, 2)
f2 = Fraction(2, 3)
print(Fraction.mul2(f1, f2))
```

```python
class MyTime:
    def __init__(self, hour, minute, second=0):
        self.hour = hour
        self.minute = minute
        self.second = second

    def __str__(self):
        return f"{self.hour:02d}:{self.minute:02d}:{self.second:02d}"


time = MyTime(1, 23)
print(time)
```

```python
class Book:
    def __init__(self, title, isbn):
        self.__title = title
        self.__isbn = isbn

    def __repr__(self):
        return f'Book("{self.__title}", "{self.__isbn}")'

    def __str__(self):
        return f"ISBN1: {self.__title}; TITLE: {self.__title}"


book = Book("The Python Tutorial", "0123456")
print(book)
print(repr(book))
```
