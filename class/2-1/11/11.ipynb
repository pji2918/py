```python
class Tv:
    # static field
    serialnum = 0

    # instance fields
    def __init__(self, on: bool, ch: int, vol: int):
        self.on = on
        self.channel = ch
        self.volume = vol
        Tv.serialnum += 1
        self.num = Tv.serialnum

    def print(self):
        print(self.on, self.channel, self.volume, self.num)


t1 = Tv(True, 11, 10)
t1.print()
t2 = Tv(True, 9, 20)
t2.print()
```

```python
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, o):
        return self.x == o.x and self.y == o.y

    def __add__(self, o):
        return Vector2D(self.x + o.x, self.y + o.y)

    def __str__(self):
        return f"({self.x}, {self.y})"


v1 = Vector2D(2, 3)
v2 = Vector2D(1, 2)
print(v1 == v2)
v3 = v1 + v2
print(v3)
```

```python
class Fraction:
    def __init__(self, ja, mo):
        self.ja = ja
        self.mo = mo

    def __add__(self, other):
        if self.mo != other.mo:
            return Fraction(
                (self.ja * other.mo) + (other.ja * self.mo), self.mo * other.mo
            )
        else:
            return Fraction(self.ja + other.ja, self.mo + other.mo)

    def __mul__(self, other):
        return Fraction(self.ja * other.ja, self.mo * other.mo)

    def __str__(self):
        return f"{self.ja}/{self.mo}"


f1 = Fraction(1, 2)
f2 = Fraction(2, 3)
print(f1 + f2)
print(f1 * f2)
```
