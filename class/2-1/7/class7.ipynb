```python
lst = list(range(5))
print(lst)

lst = list(range(5, 10))
print(lst)

lst = [1, 2, 3] + [4, 5, 6]
print(lst)

lst = [1, 2, 3] * 3
print(lst)

lst = list(range(5))
print(5 in lst)
```

```python
lst = [1, 3, 5, 7, 9]

print(len(lst))
print(max(lst))
print(min(lst))
print(sum(lst))

print(list((1, 2, 3)))

print(any([1, 2, 3]))
print(any([False, 0, ""]))

print(all([1, 2, 3]))
print(all([False, 1, ""]))
```

```python
letters = ["A", "B", "C", "D", "E", "F"]

print(letters[0])
print(letters[1])
print(letters[5])
print(letters[-1])

print(letters[0:6])
print(letters[:])
print(letters[:-1])
print(letters[::2])
print(letters[1::2])
```
