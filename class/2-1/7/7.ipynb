```python
import random


def get_var(lst: list[int], avg: float) -> float:
    sum = 0
    for i in lst:
        sum += (i - avg) ** 2
    return sum / len(lst)


n = int(input("정수 입력: "))
# lst = [random.randint(0, 100) for _ in range(n)]
lst: list[int] = []
for _ in range(n):
    lst.append(random.randint(1, 100))

avg = sum(lst) / len(lst)
# var = sum([(i - avg) ** 2 for i in lst]) / len(lst)
var = get_var(lst, avg)
print(f"AVG = {avg}")
print(f"VAR = {var}")
print(f"STD = {var**0.5}")
```
