# for 반복문


for loop를 이용하여 $n$의 팩토리얼 값 구하기

팩토리얼: $n! = 1 \times 2 \times 3 \times ... \times n$


````python
n = int(input("정수 입력: "))

fact = 1
for i in range(1, n + 1):
    fact *= i

print(f"{n}! == {fact}")
````

for loop를 이용하여 $n$의 약수 구하기


````python
n = int(input("정수 입력: "))

print(f"{n}의 약수: ", end="")

for i in range(1, n + 1):
    if n % i == 0:
        print(i, end=" ")
````

실습: 두 정수를 입력받아 3의 배수와 그 갯수 출력하기

예시

```plaintext
두 정수 입력 : 3 10
3의 배수 : 3 6 9
개수 : 3개
```


````python
x, y = map(int, input("두 정수 입력: ").split())

cnt = 0
print("3의 배수: ", end="")

for i in range(x, y + 1):
    if i % 3 == 0:
        cnt += 1
        print(i, end=" ")

print(f"\n개수: {cnt}개")
````

for문을 이용하여 원주율($\pi$)의 근삿값 구하기


````python
pi = 0

for i in range(1, 1000000):
    if i % 2 == 1:
        pi += 1 / (2 * i - 1)
    else:
        pi -= 1 / (2 * i - 1)

print("원주율 =", pi * 4)
````

````python
pi = 1

for i in range(1, 1000000):
    pi *= (4 * i**2) / (4 * i**2 - 1)

print("원주율 =", pi * 2)
````
