# for 반복문


for 반복문은 반복 횟수가 명확히 정해져 있을 때 사용한다.


for loop를 이용하여 $n$의 팩토리얼 값 구하기

팩토리얼: $n! = 1 \times 2 \times 3 \times ... \times n$


````python
n = int(input("정수 입력: "))

fact = 1
for i in range(1, n + 1):
    fact *= i

print(f"{n}! == {fact}")
````

for loop를 이용하여 $n$의 약수 구하기


````python
n = int(input("정수 입력: "))

print(f"{n}의 약수: ", end="")

for i in range(1, n + 1):
    if n % i == 0:
        print(i, end=" ")
````

실습: 두 정수를 입력받아 3의 배수와 그 갯수 출력하기

예시

```plaintext
두 정수 입력 : 3 10
3의 배수 : 3 6 9
개수 : 3개
```


````python
x, y = map(int, input("두 정수 입력: ").split())

cnt = 0
print("3의 배수: ", end="")

for i in range(x, y + 1):
    if i % 3 == 0:
        cnt += 1
        print(i, end=" ")

print(f"\n개수: {cnt}개")
````

for문을 이용하여 원주율($\pi$)의 근삿값 구하기


````python
pi = 0

for i in range(1, 1000000):
    if i % 2 == 1:
        pi += 1 / (2 * i - 1)
    else:
        pi -= 1 / (2 * i - 1)

print("원주율 =", pi * 4)
````

````python
pi = 1

for i in range(1, 1000000):
    pi *= (4 * i**2) / (4 * i**2 - 1)

print("원주율 =", pi * 2)
````

# while 반복문


while 반복문은 반복 횟수가 명확히 정해져 있지 않을 때 사용한다.


실습: 숫자 맞추기 게임


````python
import random

target = random.randint(1, 100)

userInput = 0

while target != userInput:
    userInput = int(input("숫자 입력: "))

    if userInput > target:
        print(f"{userInput}: DOWN!")
    elif userInput < target:
        print(f"{userInput}: UP!")

print(f"{userInput}: GOOD!")
````

유클리드 호제법 (최대공약수 구하기)


````python
num1, num2 = map(int, input("두 수 입력").split())

origx = min(num1, num2)
origy = max(num1, num2)

x, y = origx, origy

cnt = 0
while y > 0:
    r = x % y
    x = y
    y = r
    print(f"({x}, {y})")
    cnt += 1

print()
print(f"반복 횟수: {cnt}회")
print(f"{num1}과 {num2}의 최대공약수: {x}")
print(f"{num1}과 {num2}의 최소공배수: {int(origx * origy / x)}")
````

실습:
두 정수 a, b를 입력받아 몫과 나머지를 출력.

단, `//` `%` 사용 금지.


````python
a, b = map(int, input("두 수 입력: ").split())

cnt = 0

while a >= b:
    cnt += 1
    print(f"{cnt:2d}회: {a} - {b} = {a - b}")
    a -= b

print(f"\n몫 = {cnt}, 나머지 = {a}")
````
