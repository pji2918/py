````python
print("사전 프로그램 시작... 종료는 q를 입력")
dic = {}

while True:
    st = input("$ ")
    command = st[0]
    if command == "<":
        st = st[1:]
        inputStr = st.split(":")
        if len(inputStr) < 2:
            print("입력 오류")
        else:
            dic[inputStr[0].strip()] = inputStr[1].strip()
    elif command == ">":
        st = st[1:]
        inputStr = st.strip()
        if inputStr in dic:
            print(dic[inputStr])
        else:
            print(f"{inputStr}은 사전에 없음")
    elif command == "q":
        break
    else:
        print("입력 오류")

print("사전 프로그램 종료")
````

실습:
`lst = [1, 2, 3, 4, 5, 1, 2, 3, 1, 2]`에서 숫자의 빈도를 딕셔너리 형태로 생성하여 출력하십시오.

예시: `{1:3, 2:3, 3:2, 4:1, 5:1}`


````python
lst = [1, 2, 3, 4, 5, 1, 2, 3, 1, 2]
dic: dict[int, int] = {}

for i in lst:
    if i in dic:
        dic[i] += 1
    else:
        dic[i] = 1

print(dic)
````

실습: 영문장을 입력받아, 각 문자의 빈도 수 출력

예:

입력: `영문장 입력: hello world`

출력:

```plaintext
h : 1
e : 1
l : 3
o : 2
```


````python
sentence = input("영문장 입력: ").lower()
dic: dict[str, int] = {}

for i in sentence:
    if i != " ":
        if i in dic:
            dic[i] += 1
        else:
            dic[i] = 1

for k, v in dic.items():
    print(f"{k}({k.upper()}) : {v}개")
````

실습: 두 딕셔너리를 합쳐서 출력

예시:

```python
dict1 = {"A": 1, "C": 5}
dict2 = {"B": 2, "C": 3}
```

출력: `{"A":1, "C":8, "B": 2}`


````python
dict1 = {"A": 1, "C": 5}
dict2 = {"B": 2, "C": 3}

newDict: dict[str, int] = dict(dict1)

for key, value in dict2.items():
    if key in newDict:
        newDict[key] += value
    else:
        newDict[key] = value

print(newDict)
````
