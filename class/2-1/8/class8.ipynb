```python
print("사전 프로그램 시작... 종료는 q를 입력")
dic = {}

while True:
    st = input("$ ")
    command = st[0]
    if command == "<":
        st = st[1:]
        inputStr = st.split(":")
        if len(inputStr) < 2:
            print("입력 오류")
        else:
            dic[inputStr[0].strip()] = inputStr[1].strip()
    elif command == ">":
        st = st[1:]
        inputStr = st.strip()
        if inputStr in dic:
            print(dic[inputStr])
        else:
            print(f"{inputStr}은 사전에 없음")
    elif command == "q":
        break
    else:
        print("입력 오류")

print("사전 프로그램 종료")
```

실습:
`lst = [1, 2, 3, 4, 5, 1, 2, 3, 1, 2]`에서 숫자의 빈도를 딕셔너리 형태로 생성하여 출력하십시오.

예시: `{1:3, 2:3, 3:2, 4:1, 5:1}`


```python
lst = [1, 2, 3, 4, 5, 1, 2, 3, 1, 2]
dic: dict[int, int] = {}

for i in lst:
    if i in dic:
        dic[i] += 1
    else:
        dic[i] = 1

print(dic)
```
