연산자 우선순위


```python
x = int(input("첫 번째 수 입력 : "))
y = int(input("두 번째 수 입력 : "))
z = int(input("세 번째 수 입력 : "))

# avg = x + y + z / 3  # wrong (나누기가 먼저 계산됨)
avg = (x + y + z) / 3  # correct
print("평균 =", avg)
print("평균 = %.1f" % avg)
```

형식 지정자


```python
x = int(input("첫 번째 수 입력 : "))
y = int(input("두 번째 수 입력 : "))
z = int(input("세 번째 수 입력 : "))

avg = (x + y + z) / 3
print("평균 =", avg)
print(
    "평균 = %.1f, %.2f"  # 값의 개수와 형식 지정자 개수 맞추기
    % (avg, avg)  # 여러 개의 값을 형식 지정하여 출력할 때는 tuple로 묶기
)
```

여러 값 동시에 입력받기


```python
x, y, z = map(int, input("세 정수 입력 : ").split())

print("x=%d, y=%d, z=%d" % (x, y, z))
```

평균


```python
import math

a, b = map(int, input("두 정수 입력: ").split())

print("산술 평균 = ", (a + b) / 2)
print("기하 평균 = ", math.sqrt(a * b))  # or (a * b) ** 0.5
print("조화 평균 = ", (2 * a * b) / (a + b))
```

`%` 연산자의 용도

- 1의 자리 추출 `n % 10`
- 짝수/홀수 확인 `n % 2 == 0` => n은 짝수
- 배수/약수 확인
  - `n % 4 == 0` => n은 4의 배수
  - `n % p == 0` => n은 p의 배수
  - p가 n의 약수?


실습 - 년도 값을 입력받아 윤년/평년 출력

조건:

1. 4의 배수이면서 100의 배수가 아니어야 함
2. 혹은 400의 배수


```python
year = int(input("년도 입력: "))

print(
    f"{year}년의 윤년 여부 = {((year % 4 == 0) and (year % 100 != 0)) or (year % 400 == 0)}"
)
```
