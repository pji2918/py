#### 조건문


if-else 조건문


```python
year = int(input("연도 입력"))

if (year % 4 == 0 and year % 100 != 0) or year % 400 == 0:
    print(year, "년은 윤년")
else:
    print(year, "년은 평년")
```

```python
x, y = map(float, input("점의 좌표 x, y를 입력: ").split())

if x * x + y * y > 25:
    print("외부 점")
else:
    print("내부 점")
```

#### 중첩 조건문


else-if


```python
a = int(input("정수 입력: "))

if a <= 30:
    print("small")
elif a <= 70:
    print("middle")
else:
    print("large")
```

nested-if


```python
a = int(input("정수 입력: "))

if a > 30:
    if a > 70:
        print("large")
    else:
        print("middle")
else:
    print("small")
```

nested-if


```python
score = int(input("점수 입력: "))

if score < 90:
    if score < 80:
        if score < 70:
            if score < 60:
                print("F")
            else:
                print("D")
        else:
            print("C")
    else:
        print("B")
else:
    print("A")
```

else-if


```python
score = int(input("점수 입력: "))

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
elif score >= 60:
    print("D")
else:
    print("F")
```

```python
a, b, c = map(int, input("세 정수 입력").split())

if a > b:
    if a > c:
        print("max =", a)
    else:
        print("max =", c)
else:
    if b > c:
        print("max =", b)
    else:
        print("max =", c)
```

#### random


```python
import random

print(random.random())
print(random.randint(1, 10))
print(random.randrange(10))
print(random.randrange(1, 11))
print(random.randrange(0, 10, 2))

lst = [1, 2, 3, 4, 5]
print(random.choice(lst))
random.shuffle(lst)
print(lst)
```

```python
import math


print("math.fabs(-3.3) =", math.fabs(-3.3))
print("math.ceil(-3.3) =", math.ceil(3.3))
print("math.floor(-3.3) =", math.floor(3.9))
print("math.pow(2, 3) =", math.pow(2, 3))
print("math.log(math.e) =", math.log(math.e))
print("math.log10(10) =", math.log10(10))
print("math.sin(math.radians(90)) =", math.sin(math.radians(90)))
print("math.cos(math.radians(90)) =", math.cos(math.radians(90)))
```
