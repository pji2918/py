```python
class Shape:
    def draw(self):
        print("Shape Draw")


class Circle(Shape):
    def draw(self):
        print("Circle Draw")


c = Circle()
c.draw()
```

```python
class Shape:
    def __init__(self, name):
        self.name = name

    def getArea(self):
        pass


class Circle(Shape):
    def __init__(self, name, radius):
        super().__init__(name)
        self.radius = radius

    def getArea(self):
        return 3.14 * self.radius**2


class Rectangle(Shape):
    def __init__(self, name, width, height):
        super().__init__(name)
        self.width = width
        self.height = height

    def getArea(self):
        return self.width * self.height


shapeList = [Circle("c1", 10), Rectangle("r1", 10, 10)]
for i in shapeList:
    print(i.getArea())
```

```python
class Animal:
    def speak(self):
        pass


class Cat(Animal):
    def speak(self):
        print("야옹")


class Dog(Animal):
    def speak(self):
        print("멍멍")


lst = [Cat(), Dog()]
for i in lst:
    i.speak()
```
