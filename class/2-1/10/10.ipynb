````python
def process(w: str):
    output = ""
    for ch in w:
        if ch.isalpha():
            output += ch
    return output.lower()


fname = input("파일 이름 입력")

words = set()

with open(fname, "r") as f:
    for line in f:
        for word in line.split():
            words.add(process(word))

print(f"사용된 단어의 개수 = {len(words)}")
print(words)
````

실습: 샘플 텍스트 파일을 읽어서, 줄번호를 붙여 출력

sample.txt>

```plaintext
hello
hi
test
```

출력 예:

```plaintext
1. hello
2. hi
3. test
라인수 = 3
```


````python
with open("sample.txt") as f:
    for idx, value in enumerate(f):
        print(f"{idx + 1}. {value.strip()}")
    f.seek(0)
    print(f"라인수 = {len(f.readlines())}")
````

실습: 두 텍스트 파일을 하나의 텍스트 파일로 합치고, 결과 출력

sample1.txt>

```plaintext
hellohi
test
```

sample2.txt>

```plaintext
hello
world
```

출력:

```plaintext
hellohi
test
hello
world
```


````python
with (
    open("sample1.txt") as s1,
    open("sample2.txt") as s2,
    open("sample3.txt", "w+") as s3,
):
    s3.write(s1.read() + "\n" + s2.read())
    s3.seek(0)
    print(s3.read())
````

````python
class Tv:
    def __init__(self, on, ch, vol):
        self.on = on
        self.channel = ch
        self.volume = vol

    def print(self):
        print(self.on, self.channel, self.volume)

    def turnOn(self):
        self.on = True

    def turnOff(self):
        self.on = False

    def changeChannel(self, ch):
        self.channel = ch

    def changeVolume(self, vol):
        self.volume = vol
````

````python
t1 = Tv(False, 11, 10)
t1.turnOn()
t1.print()

t2 = Tv(False, 0, 0)
t2.turnOn()
t2.changeChannel(9)
t2.changeVolume(15)
t2.print()
````
