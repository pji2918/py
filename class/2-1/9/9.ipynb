```python
dict = {"book": "책", "note": "노트", "pen": "연필"}

sdict = sorted(dict.items(), key=lambda x: x[1])
print(sdict)
sdict = sorted(dict.items(), key=lambda x: x[0])
print(sdict)
```

```python
def square(n):
    return n * n


lst = [1, 2, 3, 4, 5]
print(list(map(square, lst)))
print(list(map(lambda n: n**2, lst)))
```

```python
def fun1(x, *args):
    print("x=", x)
    print("args=", args)


def fun2(x, **kwargs):
    print("x=", x)
    print("kargs=", kwargs)


fun1(10, 1, 2, 3)
fun2(20, a=1, b=2, c=3)
```

실습: positional variable parameter를 사용하여 (min,max) 튜플을 반환하는 함수를 작성하시오.


```python
def min_max(*args) -> tuple[int, int]:
    return (min(args), max(args))


minimum, maximum = min_max(1, 3, 9, 5, 7)
print("min =", minimum)
print("max =", maximum)

print()
minimum, maximum = min_max(1, 3, 7)
print("min =", minimum)
print("max =", maximum)
```
