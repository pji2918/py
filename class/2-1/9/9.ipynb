실습: positional variable parameter를 사용하여 (min,max) 튜플을 반환하는 함수를 작성하시오.


```python
def min_max(*args) -> tuple[int, int]:
    return (min(args), max(args))


minimum, maximum = min_max(1, 3, 9, 5, 7)
print("min =", minimum)
print("max =", maximum)

print()
minimum, maximum = min_max(1, 3, 7)
print("min =", minimum)
print("max =", maximum)
```

실습: 딕셔터리를 정렬된 리스트롤 반환하는 함수를 작성하시오.


```python
def Dsort(dic, k: bool = True, r: bool = False) -> list[tuple]:
    return sorted(dic.items(), key=lambda x: x[not k], reverse=r)


def D_item_print(d: list[tuple]) -> None:
    print("[", end="")
    for idx, value in enumerate(d):
        print(f"({value[0]},{value[1]}){', ' if not idx == len(d) - 1 else ''}", end="")
    print("]")


dic = {"banana": 3, "apple": 5, "cherry": 4}
D_item_print(Dsort(dic, True, False))
D_item_print(Dsort(dic, True, True))
D_item_print(Dsort(dic, False, r=False))
D_item_print(Dsort(dic, False, True))
```

실습: 로또 번호(1-46) 6개를 중복 없이 출력  
힌트: set(집합) 사용

예: 7, 15, 23, 39, 41, 46


```python
import random

nums: set[int] = set()

while len(nums) < 6:
    nums.add(random.randint(1, b=46))

for idx, value in enumerate(iterable=nums):
    print(f"{value}{', ' if idx < len(nums) - 1 else ''}", end="")
```
