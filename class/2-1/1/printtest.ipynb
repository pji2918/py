시:분:초 변환


```python
n = int(input("초 입력: "))
h = n // (60 * 60)
m = (n - h * 60 * 60) // 60
s = n % 60
print(h, ":", m, ":", s)
```

피타고라스의 정리


$c=\sqrt{a^2+b^2}=(a^2+b^2)^\frac{1}{2}$


```python
a = float(input("직각삼각형 밑변의 길이 입력"))
b = float(input("직각삼각형 높이 입력"))
c = (a**2 + b**2) ** 0.5
print("빗변의 길이 = ", c)
```

비트 연산


```python
print("9 = ", bin(9))
print("10 = ", bin(10))
print("9 & 10 = ", bin(9 & 10))
print("9 | 10 = ", bin(9 | 10))
print("9 ^ 10 = ", bin(9 ^ 10))
```

```python
print("9 << 1 = ", 9 << 1)
print("9 >> 1 = ", 9 >> 1)
print("~9 = ", ~9)
```

```python
print("12 = ", bin(12))
print("11 = ", bin(11))
print("12 & 11 = ", bin(12 & 11), "=", 12 & 11)
print("12 | 11 = ", bin(12 | 11), "=", 12 | 11)
print("12 ^ 11 = ", bin(12 ^ 11), "=", 12 ^ 11)
```

```python
print("12 << 1 = ", 12 << 1)
print("12 >> 1 = ", 12 >> 1)
print("11 << 1 = ", 11 << 1)
print("11 >> 1 = ", 11 >> 1)
print("~12 = ", ~12)
print("~11 = ", ~11)
```

**Key Insight**

Left Shift: << 1 == x \* 2
<br/>Right Shift: >> 1 == x / 2
<br/>NOT: ~x == -x-1
