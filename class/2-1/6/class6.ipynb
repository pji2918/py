```python
def order(num, pickle=True, onion=True):
    print(f"버거 {num}개: 피클 {pickle}, 양파 {onion}")


def power(base, exp):
    return base**exp


order(1)
order(2, False)
order(2, False, False)
print(pow(2, 3))
print(pow(exp=3, base=2))
```

```python
def weeklyPay(rate: int, hour: int) -> float:
    if hour > 30:
        money: float = rate * 30 + 1.5 * rate * (hour - 30)
    else:
        money: float = rate * hour
    return money


r = int(input("시급 입력: "))
h = int(input("근무 시간 입력: "))
print("주급은 " + str(weeklyPay(rate=r, hour=h)))
```

$n! = n \times (n - 1)!$


```python
def factorial(n: int) -> int:
    if n <= 1:
        return 1
    else:
        return n * factorial(n - 1)


print("5! =", factorial(5))
```

```python
def fib(i):
    if i == 1:
        return 0
    elif i == 2:
        return 1
    else:
        return fib(i - 1) + fib(i - 2)


def fib2(i):
    if i <= 1:
        return 0
    elif i <= 2:
        return 1
    else:
        nums = [0, 1]
        for _ in range(i - 2):
            nums.append(nums[-1] + nums[-2])
        return nums[-1]


def fib3(i):
    n1 = 0
    n2 = 1
    sum = 0
    for _ in range(1, i):
        n1 = n2
        n2 = sum
        sum = n1 + n2
    return sum


n = int(input("정수 입력 : "))
# print("%d번째 피보나치 수열 = %d" % (n, fib(n)))
# print("%d번째 피보나치 수열 = %d" % (n, fib2(n)))

for i in range(1, n + 1):
    print(fib2(i), end=" ")
print()
for i in range(1, n + 1):
    print(fib3(i), end=" ")
```

```python
def rsum(n):
    if n == 1:
        return 1
    else:
        return 1 / n + rsum(n - 1)


n = int(input("정수 입력: "))
print(f"1부터 {n}까지의 역수 수열의 합: {rsum(100):.1f}")
```

```python
def sumAB(a, b):
    if b <= a:
        return a
    else:
        return b + sumAB(a, b - 1)


n1, n2 = map(int, input("두 정수 입력: ").split())
print(sumAB(n1, n2))
```
