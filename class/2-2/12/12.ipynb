```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn import linear_model

regr = linear_model.LinearRegression()

X = [[164], [179], [162], [170]]
y = [53, 63, 55, 59]
regr.fit(X, y)

plt.scatter(X, y, color="black")

y_pred = regr.predict(X)

plt.plot(X, y_pred, color="blue", linewidth=3)
plt.show()

print(f"SCORE = {regr.score(X, y)}")
```

```python
import numpy as np
import pandas as pd
import tensorflow as tf

파일경로 = (
    "https://raw.githubusercontent.com/blackdew/tensorflow1/master/csv/lemonade.csv"
)
레모네이드 = pd.read_csv(파일경로)

독립 = 레모네이드[["온도"]]
종속 = 레모네이드[["판매량"]]
print(독립.shape, 종속.shape)

X = tf.keras.layers.Input(shape=[1])
Y = tf.keras.layers.Dense(1)(X)
model = tf.keras.models.Model(X, Y)
model.compile(loss="mse")

model.fit(독립, 종속, epochs=3000, verbose=0)
model.fit(독립, 종속, epochs=10)

print(model.predict(독립))
print(model.predict(np.array([[15]])))
```

```python
아이리스 = pd.read_csv("iris.csv")
인코딩 = pd.get_dummies(아이리스)

독립 = 인코딩[["sepal_length", "sepal_width", "petal_length", "petal_width"]]
종속 = 인코딩[
    ["species_Iris-setosa", "species_Iris-versicolor", "species_Iris-virginica"]
]
print(독립.shape, 종속.shape)

X = tf.keras.layers.Input(shape=[4])
Y = tf.keras.layers.Dense(3, activation="softmax")(X)
model = tf.keras.models.Model(X, Y)
model.compile(loss="categorical_crossentropy", metrics=["accuracy"])

model.fit(독립, 종속, epochs=100)

print(model.predict(독립[:5]))
print(종속[:5])
```

```python
import matplotlib.pyplot as plt
import numpy as np
import tensorflow as tf
from tensorflow import keras
```

```python
mnist = keras.datasets.mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
plt.imshow(x_train[0], cmap="gray")
plt.show()
```

```python
x_train = x_train / 255
x_test = x_test / 255
```

```python
for data in x_train[46]:
    for i in data:
        print(f"{i:^.1f}", end="   ")
    print()
```

```python
model = keras.models.Sequential(
    [
        keras.layers.Flatten(input_shape=[28, 28]),
        keras.layers.Dense(200, activation="relu"),
        keras.layers.Dense(100, activation="relu"),
        keras.layers.Dense(50, activation="relu"),
        keras.layers.Dense(10, activation="softmax"),
    ]
)
```

```python
model.compile(
    optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"]
)
```

```python
model.fit(x_train, y_train, epochs=5)
```

```python
model.evaluate(x_test, y_test)
```

```python
pred = model.predict(x_test)
print(type(pred), pred.shape)
print(model.predict(x_test).round(2))
```

```python
plt.imshow(x_test[46], cmap="gray")
plt.show()

pred = model.predict(x_test[46].reshape(1, 28, 28))
print(type(pred), pred.shape)
print(pred.round(2))
print(np.argmax(pred), y_test[46])
```
