```python
import time

import numpy as np

arr_a = np.random.rand(10000)
arr_b = np.random.rand(10000)
result = np.zeros(10000)

start = time.time()
for i in range(len(arr_a)):
    result[i] = arr_a[i] * arr_b[i]
end = time.time()

time_a = end - start
print("소요시간:", time_a)
```

```python
start = time.time()
result = arr_a * arr_b
end = time.time()

time_b = end - start

print("소요시간:", time_b)
print(f"{(time_a / time_b)}배 차이")
```

```python
import time

import numpy as np

n = 1000
arr = np.random.rand(n, n)

sum_native = 0

start = time.time()
for i in range(n):
    for j in range(n):
        sum_native += arr[i, j]
end = time.time()

time_a = end - start

print(f"{n}*{n} 배열 원소의 합: {sum_native}")
print(f"소요시간: {time_a}초")
```

```python
start = time.time()
sum_reduction = np.sum(arr, (0, 1))
end = time.time()

time_b = end - start

print(f"{n}*{n} 배열 원소의 합: {sum_reduction}")
print(f"소요시간: {time_b}초")
print(f"{(time_a / time_b)}배 차이")
```
