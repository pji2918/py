```python
import numpy as np

arr = np.array([[1, 2, 3], [4, 5, 6]])
larr = arr % 2 == 0
print(larr)
print(arr[larr])
```

```python
import numpy as np

arr = np.array([[1, 2, 3], [4, 5, 6]])

larr = arr[:, 0] % 2 == 0
print(arr[:, 0], larr)
print(arr[larr])
```

```python
import numpy as np

lst = [
    [177, 77.1],
    [183, 80.1],
    [185, 78.3],
    [173, 80.3],
]
arr = np.array(lst)

print("몸무게가 80 이상")
print(arr[arr[:, 1] >= 80])

print("키가 180 이상")
print(arr[arr[:, 0] >= 180])
```

```python
arr1 = np.arange(8).reshape(2, 4)
arr2 = np.arange(8, 16).reshape(2, 4)

print(arr1)
print(arr2)
print(np.concatenate((arr1, arr2), axis=0))
print(np.concatenate((arr1, arr2), axis=1))
```

```python
arr = np.arange(8).reshape(2, 4)
print(arr)

up, down = np.split(arr, 2, axis=0)
print(up)
print(down)

left, right = np.split(arr, 2, axis=1)
print(left)
print(right)
```
