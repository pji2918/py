```python
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def __add__(self, other):
        return Vector2D(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __repr__(self):
        return f"Vector2D({self.x}, {self.y})"


v1 = Vector2D(2, 3)
v2 = Vector2D(1, 2)
print(v1 == v2)
v3 = v1 + v2
print(v3)

repr(v3)
```

```python
class Fraction:
    def __init__(self, ja, mo):
        self.ja = ja
        self.mo = mo

    def __add__(self, other):
        if self.mo == other.mo:
            return Fraction(self.ja + other.ja, self.mo)
        else:
            return Fraction(
                (self.ja * other.mo) + (other.ja * self.mo), self.mo * other.mo
            )

    def __mul__(self, other):
        return Fraction(self.ja * other.ja, self.mo * other.mo)

    def __str__(self):
        return f"{self.ja}/{self.mo}"


f1 = Fraction(1, 2)
f2 = Fraction(2, 3)

print(f1 + f2)
print(f1 * f2)
```

```python
class User:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def from_string(cls, user_str):
        name, age = user_str.split(",")
        return cls(name, age)


user = User.from_string("홍길동,30")

print(user.name, user.age)
```

```python
class Fraction:
    def __init__(self, ja, mo):
        self.ja = ja
        self.mo = mo

    def __str__(self):
        return f"{self.ja}/{self.mo}"

    @classmethod
    def mul2(cls, f1, f2):
        return cls(f1.ja * f2.ja, f1.mo * f2.mo)


f1 = Fraction(1, 2)
f2 = Fraction(2, 3)

print(Fraction.mul2(f1, f2))
```

```python
class Shape:
    def __init__(self, name):
        self.name = name

    def getArea(self):
        pass


class Circle(Shape):
    def __init__(self, name, radius):
        super().__init__(name)
        self.radius = radius

    def getArea(self):
        return 3.14 * self.radius**2


class Rectangle(Shape):
    def __init__(self, name, width, height):
        super().__init__(name)
        self.width = width
        self.height = height

    def getArea(self):
        return self.width * self.height


shapeList = [Circle("c1", 10), Rectangle("r1", 10, 10)]

for s in shapeList:
    print(s.getArea())
```
