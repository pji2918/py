```python
class Tv:
    def __init__(self, on, channel, volume):
        self.on = on
        self.channel = channel
        self.volume = volume

    def print(self):
        print(self.on, self.channel, self.volume)

    def turnOn(self):
        self.on = True

    def turnOff(self):
        self.on = False

    def changeVolume(self, volume):
        self.volume = volume

    def changeChannel(self, channel):
        self.channel = channel


t1 = Tv(False, 11, 10)
t1.turnOn()
t1.print()

t2 = Tv(False, 0, 0)
t2.turnOn()
t2.changeChannel(9)
t2.changeVolume(15)
t2.print()
```

```python
class Tv:
    serialnum = 0

    def __init__(self, on, channel, volume):
        self.on = on
        self.channel = channel
        self.volume = volume
        Tv.serialnum += 1  # ty: ignore[unresolved-attribute]
        self.num = Tv.serialnum  # ty: ignore[unresolved-attribute]

    def print(self):
        print(self.on, self.channel, self.volume, self.num)

    def turnOn(self):
        self.on = True

    def turnOff(self):
        self.on = False

    def changeVolume(self, volume):
        self.volume = volume

    def changeChannel(self, channel):
        self.channel = channel


t1 = Tv(True, 11, 10)
t1.print()

t2 = Tv(True, 9, 20)
t2.print()
```

```python
class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def __add__(self, other):
        return Vector2D(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __repr__(self):
        return f"Vector2D({self.x}, {self.y})"


v1 = Vector2D(2, 3)
v2 = Vector2D(1, 2)
print(v1 == v2)
v3 = v1 + v2
print(v3)

repr(v3)
```

```python
class Fraction:
    def __init__(self, ja, mo):
        self.ja = ja
        self.mo = mo

    def __add__(self, other):
        if self.mo == other.mo:
            return Fraction(self.ja + other.ja, self.mo)
        else:
            return Fraction(
                (self.ja * other.mo) + (other.ja * self.mo), self.mo * other.mo
            )

    def __mul__(self, other):
        return Fraction(self.ja * other.ja, self.mo * other.mo)

    def __str__(self):
        return f"{self.ja}/{self.mo}"


f1 = Fraction(1, 2)
f2 = Fraction(2, 3)

print(f1 + f2)
print(f1 * f2)
```
