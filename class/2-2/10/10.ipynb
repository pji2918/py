```python
import pandas as pd

data = {
    "Product": [
        "Apple",
        "Banana",
        "Apple",
        "Orange",
        "Banana",
        "Orange",
        "Apple",
    ],
    "Category": ["Fruit", "Fruit", "Fruit", "Fruit", "Fruit", "Fruit", "Fruit"],
    "Store": ["Seoul", "Seoul", "Busan", "Seoul", "Busan", "Busan", "Seoul"],
    "Sales": [1500, 800, 1200, 2000, 900, 1500, 3000],
    "Quantity": [5, 10, 4, 8, 12, 6, 9],
}

df = pd.DataFrame(data)

print("--- [원본 데이터] ---")
print(df, "\n")

seoul_df: pd.DataFrame = df[df["Store"] == "Seoul"]
print("--- [1. 서울 지점 데이터 필터링] ---")
print(seoul_df, "\n")

sorted_df = seoul_df.sort_values(by="Sales", ascending=False)
print("--- [2. 서울 지점 매출 기준 내림차순 정렬] ---")
print(sorted_df, "\n")

grouped_df = (
    df.groupby(by="Product").agg({"Sales": "sum", "Quantity": "mean"}).reset_index()
)
print("--- [3. 상품별 매출 총합 및 평균 판매 수량] ---")
print(grouped_df)
```

```python
ages = [15, 22, 25, 29, 31, 46, 52, 65]
df = pd.DataFrame({"Ages": ages})
bins = [0, 19, 35, 60, 100]
labels = ["Minor", "Youth", "Middle", "Senior"]
df["Age_Group_Cut"] = pd.cut(df["Ages"], bins=bins, labels=labels)
df["Age_Group_Qcut"] = pd.qcut(df["Ages"], q=4, labels=[f"Q{x}" for x in range(1, 5)])
print(df)
```

```python
import pandas as pd

data = {
    "직원ID": [1, 2, 3, 4],
    "부서": ["인사", "개발", "디자인", "개발"],
    "직급": ["사원", "대리", "과장", "사원"],
}
df = pd.DataFrame(data)
print("--- 원본 데이터 ---")
print(df)

df_one_hot = pd.get_dummies(df, columns=["부서"], dtype=int)
print("\n--- '부서' 원-핫 인코딩 결과 ---")
print(df_one_hot)

rank_mapping = {"사원": 0, "대리": 1, "과장": 2}
df_ordinal = df.copy()
df_ordinal["직급_encoded"] = df_ordinal["직급"].map(rank_mapping)

print("\n--- '직급' 오디널 인코딩 결과 ---")
print(df_ordinal[["직원ID", "직급", "직급_encoded"]])
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

np.random.seed(42)
data = np.random.normal(loc=50, scale=15, size=100)
df = pd.DataFrame(data, columns=["Original"])

min_max_scaler = MinMaxScaler()
df["MinMax"] = min_max_scaler.fit_transform(df[["Original"]])

std_scaler = StandardScaler()
df["Standard"] = std_scaler.fit_transform(df[["Original"]])

print(df.describe())
```
