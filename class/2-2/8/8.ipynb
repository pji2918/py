```python
import requests
from bs4 import BeautifulSoup

url = "https://books.toscrape.com"
response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")
main_title = soup.find("a", string=True)
print(f"메인 타이틀: {main_title.text.strip()}")  # ty: ignore[unresolved-attribute]

first_book = soup.find("article", class_="product_pod")
if first_book:
    book_link = first_book.find("h3").find("a")  # ty: ignore[unresolved-attribute]
    book_title = book_link["title"]  # ty: ignore[not-subscriptable]
    book_price = first_book.find("p", class_="price_color").text  # ty: ignore[unresolved-attribute]

    print(f"책 제목: {book_title}")
    print(f"책 가격: {book_price}")
else:
    print("책 정보를 찾을 수 없습니다. URL이나 HTML 태그 구조를 확인해 주세요.")
```
