```python
import requests
from bs4 import BeautifulSoup

url = "https://books.toscrape.com"
response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")
main_title = soup.find("a", string=True)
print(f"메인 타이틀: {main_title.text.strip()}")  # ty: ignore[unresolved-attribute]

first_book = soup.find("article", class_="product_pod")
if first_book:
    book_link = first_book.find("h3").find("a")  # ty: ignore[unresolved-attribute]
    book_title = book_link["title"]  # ty: ignore[not-subscriptable]
    book_price = first_book.find("p", class_="price_color").text  # ty: ignore[unresolved-attribute]

    print(f"책 제목: {book_title}")
    print(f"책 가격: {book_price}")
else:
    print("책 정보를 찾을 수 없습니다. URL이나 HTML 태그 구조를 확인해 주세요.")
```

```python
import requests
from bs4 import BeautifulSoup

url = "https://books.toscrape.com"
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

all_books = soup.select("article.product_pod")

print(f"--- select() 결과 (총 {len(all_books)}개의 책 발견) ---")
for index, book in enumerate(all_books):
    title = book.select_one("h3 > a")["title"]  # ty: ignore[not-subscriptable]
    price = book.select_one("p.price_color").text  # ty: ignore[unresolved-attribute]

    print(f"{index}. {title} - {price}")
```

```python
import requests
from bs4 import BeautifulSoup

html_doc = """
<html>
    <body>
        <div class="container">
            <a href="/p/1" class="item">노트북</a>
        </div>
    </body>
</html>
"""

soup = BeautifulSoup(html_doc, "html.parser")
link = soup.select_one("a.item")
if link is not None:
    print(link.get("href"))
    print(link.get("title", "없음"))
    print(link.attrs)
    print(link.get_text())
```

```python
import re

from bs4 import BeautifulSoup

html = """
<ul id="notices">
    <li><a href="/notice/101">수강신청 안내</a></li>
    <li><a href="/notice/102">국가장학금 신청 안내</a></li>
    <li><a href="/notice/103">성적우수 장학금 지급 안내</a></li>
</ul>
"""

soup = BeautifulSoup(html, "html.parser")
board = soup.select_one("ul#notices")
assert board is not None
links = board.find_all("a", href=True, string=re.compile("장학"))
for i in links:
    print(f"{i.get_text()} -> {i.get('href')}")
```

```python
from bs4 import BeautifulSoup

html = """
<table>
    <tr>
        <th>가격</th>
        <td>29,000원</td>
    </tr>
</table>
"""

soup = BeautifulSoup(html, "html.parser")

header = soup.find("th", string="가격")
price = header.find_next_sibling("td").get_text()  # ty: ignore[unresolved-attribute]
print(price)
```

```python
from urllib.request import urlopen

from bs4 import BeautifulSoup

with urlopen("https://music.bugs.co.kr/chart?wl_ref=M_left_02_01") as fr:
    soup = BeautifulSoup(fr, "html.parser")
    musics = soup.find_all("tr", class_="check")

    for i, music in enumerate(musics):
        print(f"{i + 1}위: {music.input.get('title')}")
```
