```python
import pandas as pd

s1 = pd.Series(
    [9904312, 3448737, 2890451, 2466052],
    index=["서울", "부산", "인천", "대구"],
)

s1.name = "인구"
s1.index.name = "도시"

print(s1, "\n")
print("대구 인구 = ", s1.대구)

s2 = pd.Series(range(10, 14))
print(s2)
```

```python
import pandas as pd

s = pd.Series(
    [9904312, 3448737, 2890451, 2466052],
    index=["서울", "부산", "인천", "대구"],
)

print(s.iloc[3], s["대구"])
print(s.iloc[[0, 3]], "\n", s[["서울", "대구"]])
print(s[(2_500_000 < s) & (s < 5_000_000)])
```
