```python
import numpy as np
import pandas as pd

df = pd.DataFrame(
    np.arange(10, 22).reshape(3, 4),
    index=["r1", "r2", "r3"],
    columns=["A", "B", "C", "D"],
)

print(df)
print("-" * 20)
print(df.loc["r2"])
print("-" * 20)
print(df.loc["r2":"r3"])
```

```python
import numpy as np
import pandas as pd

df = pd.DataFrame(
    np.arange(10, 22).reshape(3, 4),
    index=["r1", "r2", "r3"],
    columns=["A", "B", "C", "D"],
)
print(df.loc[:, :])
print("-" * 20)
print(df.loc["r2":"r3", "B"])
print("-" * 20)
print(df.loc["r2", "B":"C"])
print("-" * 20)
print(df.loc["r2", "B"])
```

```python
import numpy as np
import pandas as pd

df = pd.DataFrame(
    np.arange(10, 22).reshape(3, 4),
    index=["r1", "r2", "r3"],
    columns=["A", "B", "C", "D"],
)
print(df)
print("-" * 20)
print(df.loc[df.A > 10, "B"])
print("-" * 20)
print(df.loc[df.A > 10, "B":"C"])
```
