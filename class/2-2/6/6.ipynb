```python
import matplotlib.pyplot as plt

x = [x for x in range(20)]
y = [x**2 for x in range(20)]
z = [x**3 for x in range(20)]

plt.plot(x, x, label="linear")
plt.plot(x, y, label="quadratic")
plt.plot(x, z, label="qubic")

plt.xlabel("x label")
plt.ylabel("y label")
plt.legend()
plt.show()
```

```python
import math

import matplotlib.pyplot as plt

x = []
y = []

for angle in range(360):
    x.append(angle)
    y.append(math.sin(math.radians(angle)))

plt.plot(x, y)

plt.title("SINE WAVE")
plt.show()
```

```python
import math

import matplotlib.pyplot as plt

angle = range(360)
x = [x for x in angle]
sin_y = [math.sin(math.radians(y)) for y in angle]
cos_y = [math.cos(math.radians(y)) for y in angle]

# for angle in range(360):
#     x.append(angle)
#     y.append(math.sin(math.radians(angle)))

plt.plot(x, sin_y, label="sin()")
plt.plot(x, cos_y, label="cos()")

plt.title("SINE/COSINE WAVE")
plt.xlabel("x: degree")
plt.ylabel("y: sin/cos(x)")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

years = [1950, 1960, 1970, 1980, 1990, 2000, 2010]
gdp = [67.0, 80.0, 257.0, 1686.0, 6505, 11865.3, 22105.3]

plt.bar(range(len(years)), gdp)

plt.title("GDP per capita income")
plt.ylabel("dollars")

plt.xticks(range(len(years)), years)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

years = [1965, 1975, 1985, 1995, 2005, 2015]
ko = [130, 650, 2450, 11600, 17790, 27250]
jp = [890, 5120, 11500, 42130, 40560, 38780]
cn = [100, 200, 290, 540, 1760, 7940]

x = np.arange(len(years))
plt.bar(x, ko, width=0.25)
plt.bar(x + 0.3, jp, width=0.25)
plt.bar(x + 0.6, cn, width=0.25)

plt.xticks(x, years)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

mu1, sigma1 = 10, 2
mu2, sigma2 = -6, 3

standard_Gauss = np.random.randn(10000)
Gauss1 = mu1 + sigma1 * np.random.randn(10000)
Gauss2 = np.random.normal(mu2, sigma2, 10000)
plt.figure(figsize=(10, 6))
plt.hist(standard_Gauss, bins=50, alpha=0.4)
plt.hist(Gauss1, bins=50, alpha=0.4)
plt.hist(Gauss2, bins=50, alpha=0.4)

plt.show()
```
