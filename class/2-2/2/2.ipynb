```python
import random

n = int(input("n: "))
lst = [random.randint(1, 100) for _ in range(n)]
avg = sum(lst) / len(lst)

var = 0
for i in lst:
    var += (i - avg) ** 2
var /= len(lst)
std = var**0.5

print(lst)
print(f"평균: {avg}")
print(f"분산: {var}")
print(f"표준편차: {std}")
```
