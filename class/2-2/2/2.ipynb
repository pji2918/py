```python
import random

n = int(input("n: "))
lst = [random.randint(1, 100) for _ in range(n)]
avg = sum(lst) / len(lst)

var = 0
for i in lst:
    var += (i - avg) ** 2
var /= len(lst)
std = var**0.5

print(lst)
print(f"평균: {avg}")
print(f"분산: {var}")
print(f"표준편차: {std}")
```

```python
print("사전 프로그램 시작... 종료는 q를 입력")

dic = {}

while True:
    prompt = input("$ ")
    if prompt[0] == "<":
        inputStr = prompt[1:].split(":")
        if len(inputStr) < 2:
            print("입력 오류")
        else:
            dic[inputStr[0].strip()] = inputStr[1].strip()
    elif prompt[0] == ">":
        inputStr = prompt[1:].strip()
        if inputStr in dic:
            print(dic[inputStr])
        else:
            print(f"{inputStr}는 사전에 없음")
    elif prompt[0] == "q":
        break
```

```python
def get_divisors(num):
    divisors = set()
    for i in range(1, num + 1):
        if num % i == 0:
            divisors.add(i)
    return divisors


x = 48
print(f"{x}의 약수: {get_divisors(x)}")
y = 60
print(f"{y}의 약수: {get_divisors(y)}")

A = get_divisors(x)
B = get_divisors(y)
print(f"{x} {y}의 공약수: {A & B}")
print(f"{x} {y}의 최대공약수: {max(A & B)}")
```
