```python
score = int(input("점수 입력:"))

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
elif score >= 60:
    print("D")
else:
    print("F")
```

```python
import math

a, b, c = map(int, input("입력: ").split())

if b**2 - (4 * a * c) > 0:
    print(
        f"두 실근: {(-b - math.sqrt(b**2 - (4 * a * c))) / (2 * a):.0f}, {(-b + math.sqrt(b**2 - (4 * a * c))) / (2 * a):.0f}"
    )
elif b**2 - (4 * a * c) == 0:
    print(f"중근: {(-b + math.sqrt(b**2 - (4 * a * c))) / (2 * a):.0f}")
else:
    print("근이 없음")
```
