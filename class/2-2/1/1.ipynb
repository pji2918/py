```python
score = int(input("점수 입력:"))

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
elif score >= 60:
    print("D")
else:
    print("F")
```

```python
import math

# 입력된 세 값을 a,b,c에 할당
a, b, c = map(int, input("입력: ").split())

# 판별식 D = b**2 - (4 * a * c)
# D > 0
if b**2 - (4 * a * c) > 0:
    print(
        f"두 실근: {(-b - math.sqrt(b**2 - (4 * a * c))) / (2 * a):.0f}, {(-b + math.sqrt(b**2 - (4 * a * c))) / (2 * a):.0f}"
    )
# D == 0
elif b**2 - (4 * a * c) == 0:
    print(f"중근: {(-b + math.sqrt(b**2 - (4 * a * c))) / (2 * a):.0f}")
# D < 0
else:
    print("근이 없음")
```

```python
def isPrime(n):
    if n == 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True


cnt = 0
for i in range(1, 1001):
    if isPrime(i):
        print(f"{i:3d}", end=" ")
        cnt += 1
        if cnt % 10 == 0:
            print()
```
