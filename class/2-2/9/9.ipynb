```python
import pandas as pd

jan = pd.DataFrame(
    {
        "order_id": ["0001", "0002"],
        "customer_id": ["C01", "C02"],
        "amount": [30000, 50000],
    }
)

feb = pd.DataFrame(
    {
        "order_id": ["0003", "0004", "0005", "0006"],
        "customer_id": ["C01", "C99", "C03", "C01"],
        "amount": [20000, 40000, 12000, 39800],
    }
)

orders = pd.concat([jan, feb], ignore_index=True, axis=0)
print(orders)
```

```python
customers = pd.DataFrame(
    {
        "customer_id": ["C01", "C02", "C03"],
        "customer_name": ["김민수", "이서연", "박지훈"],
        "region": ["서울", "부산", "대구"],
    }
)

result = orders.merge(
    customers,
    on="customer_id",
    how="left",
    validate="many_to_one",
)
print(result)
```

```python
result = orders.merge(
    customers,
    on="customer_id",
    how="right",
    validate="many_to_one",
)
print(result)
```

```python
result = orders.merge(
    customers,
    on="customer_id",
    how="inner",
    validate="many_to_one",
)
print(result)
```

```python
result = orders.merge(
    customers,
    on="customer_id",
    how="outer",
    validate="many_to_one",
)
print(result)
```

```python
a = pd.DataFrame({"이름": ["민수", "서연"]}, index=[0, 1])
b = pd.DataFrame({"점수": [90, 80]}, index=[1, 2])

print(pd.concat([a, b], axis=1))

matched_orders = orders.merge(
    customers,
    on="customer_id",
    how="inner",
    validate="many_to_one",
)

region_sales = matched_orders.groupby("region", as_index=False)["amount"].sum()
print(region_sales)
```
