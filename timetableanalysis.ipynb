Import libraries and table


```python
import pandas as pd

table = pd.read_excel(
    "~/Downloads/tt.xlsx",
    index_col=0,
    header=None,
    sheet_name=input("Enter a Sheet Name"),
).T.reset_index(drop=True)

newColIndex = []

for idx, value in enumerate(table.columns):
    if pd.isna(value):
        newColIndex.append(table.columns[idx - 1].strip() + " 출발")
    elif value not in ["시발역", "종착역", "열차번호", "연계열번"]:
        newColIndex.append(value.strip() + " 도착")
    else:
        newColIndex.append(value.strip())

table.columns = newColIndex
```

(Optional) Import additional table to compare


```python
compareTable = pd.read_excel(
    "~/Downloads/tt-1.xlsx",
    index_col=0,
    header=None,
    sheet_name=input("Enter a Sheet Name"),
).T.reset_index(drop=True)

newColIndex = []

for idx, value in enumerate(compareTable.columns):
    if pd.isna(value):
        newColIndex.append(compareTable.columns[idx - 1].strip() + " 출발")
    elif value not in ["시발역", "종착역", "열차번호", "연계열번"]:
        newColIndex.append(value.strip() + " 도착")
    else:
        newColIndex.append(value.strip())

compareTable.columns = newColIndex
```

Compare Gyeongsan Station's Timetable


```python
print(
    pd.DataFrame(
        [table["경산 출발"], compareTable["경산 출발"]],
        ["NEW", "OLD"],
    )
    .dropna(how="all", axis=1)
    .to_string(header=False)
)
```

Compare Daegu Station's Timetable


```python
print(
    pd.DataFrame(
        [
            table["대구 출발"][table["종착역"] != "동대구"],
            compareTable["대구 출발"][table["종착역"] != "동대구"],
        ],
        ["NEW", "OLD"],
    )
    .dropna(how="all", axis=1)
    .to_string(header=False)
)
```
