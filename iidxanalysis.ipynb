```python
df1_sampled_data: pd.DataFrame = df1[
    [
        "NAME",
        "BEGINNER LEVEL",
        "BEGINNER CLEAR TYPE",
        "BEGINNER DJ LEVEL",
        "NORMAL LEVEL",
        "NORMAL CLEAR TYPE",
        "NORMAL DJ LEVEL",
        "HYPER LEVEL",
        "HYPER CLEAR TYPE",
        "HYPER DJ LEVEL",
        "ANOTHER LEVEL",
        "ANOTHER CLEAR TYPE",
        "ANOTHER DJ LEVEL",
        "LEGGENDARIA LEVEL",
        "LEGGENDARIA CLEAR TYPE",
        "LEGGENDARIA DJ LEVEL",
    ]
]
melted_data: pd.DataFrame = df1_sampled_data.melt(
    id_vars="NAME",
    value_vars=[
        "BEGINNER CLEAR TYPE",
        "NORMAL CLEAR TYPE",
        "HYPER CLEAR TYPE",
        "ANOTHER CLEAR TYPE",
        "LEGGENDARIA CLEAR TYPE",
    ],
    var_name="PATTERN",
    value_name="CLEAR TYPE",
)

melted_data["PATTERN"] = melted_data["PATTERN"].str.replace(
    " CLEAR TYPE", "", regex=False
)

KEY = ("BEGINNER", "NORMAL", "HYPER", "ANOTHER", "LEGGENDARIA")

melted_data.sort_values(
    by=["NAME", "PATTERN"],
    key=lambda s: s.map(lambda x: KEY.index(x)) if s.name == "PATTERN" else s,
)
```

```python
from pandas.core.frame import DataFrame

df1_sampled_data: pd.DataFrame = df1[
    [
        "タイトル",
        "BEGINNER 難易度",
        "BEGINNER クリアタイプ",
        "BEGINNER DJ LEVEL",
        "NORMAL 難易度",
        "NORMAL クリアタイプ",
        "NORMAL DJ LEVEL",
        "HYPER 難易度",
        "HYPER クリアタイプ",
        "HYPER DJ LEVEL",
        "ANOTHER 難易度",
        "ANOTHER クリアタイプ",
        "ANOTHER DJ LEVEL",
        "LEGGENDARIA 難易度",
        "LEGGENDARIA クリアタイプ",
        "LEGGENDARIA DJ LEVEL",
    ]
]
melted_data2: pd.DataFrame = df1_sampled_data.melt(
    id_vars="NAME",
    value_vars=[
        "BEGINNER LEVEL",
        "NORMAL LEVEL",
        "HYPER LEVEL",
        "ANOTHER LEVEL",
        "LEGGENDARIA LEVEL",
    ],
    var_name="PATTERN",
    value_name="LEVEL",
)

melted_data2["PATTERN"] = melted_data2["PATTERN"].str.replace(" LEVEL", "", regex=False)

merged_data: DataFrame = melted_data.merge(
    right=melted_data2[melted_data2["LEVEL"] != 0],
    on=["タイトル", "PATTERN"],
    how="inner",
)

merged_data[merged_data["CLEAR TYPE"] != "NO PLAY"].sample(10)
```
