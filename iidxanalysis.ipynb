```python
from pathlib import Path

import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
```

## CSV 불러오기


```python
# 여기를 수정
file1 = Path("./data/1.csv")
file2 = Path("./data/2.csv")

df1 = pd.read_csv(file1)
df2 = pd.read_csv(file2)

MAPPER = {
    "バージョン": "VERSION",
    "タイトル": "NAME",
    "ジャンル": "GENRE",
    "アーティスト": "ARTIST",
    "プレー回数": "PLAY COUNT",
    "BEGINNER 難易度": "BEGINNER LEVEL",
    "BEGINNER スコア": "BEGINNER SCORE",
    "BEGINNER ミスカウント": "BEGINNER MISS COUNT",
    "BEGINNER クリアタイプ": "BEGINNER CLEAR TYPE",
    "NORMAL 難易度": "NORMAL LEVEL",
    "NORMAL スコア": "NORMAL SCORE",
    "NORMAL ミスカウント": "NORMAL MISS COUNT",
    "NORMAL クリアタイプ": "NORMAL CLEAR TYPE",
    "HYPER 難易度": "HYPER LEVEL",
    "HYPER スコア": "HYPER SCORE",
    "HYPER ミスカウント": "HYPER MISS COUNT",
    "HYPER クリアタイプ": "HYPER CLEAR TYPE",
    "ANOTHER 難易度": "ANOTHER LEVEL",
    "ANOTHER スコア": "ANOTHER SCORE",
    "ANOTHER ミスカウント": "ANOTHER MISS COUNT",
    "ANOTHER クリアタイプ": "ANOTHER CLEAR TYPE",
    "LEGGENDARIA 難易度": "LEGGENDARIA LEVEL",
    "LEGGENDARIA スコア": "LEGGENDARIA SCORE",
    "LEGGENDARIA ミスカウント": "LEGGENDARIA MISS COUNT",
    "LEGGENDARIA クリアタイプ": "LEGGENDARIA CLEAR TYPE",
    "最終プレー日時": "LAST PLAYED",
}

df1 = df1.rename(columns=MAPPER)
df2 = df2.rename(columns=MAPPER)
```

```python
merged_data: list[pd.DataFrame] = []
```

df1 내의 같은 곡 다른 패턴을 CLEAR TYPE과 함께 분리


```python
for df in (df1, df2):
    melted_data = df.melt(
        id_vars="NAME",
        value_vars=[
            "BEGINNER CLEAR TYPE",
            "NORMAL CLEAR TYPE",
            "HYPER CLEAR TYPE",
            "ANOTHER CLEAR TYPE",
            "LEGGENDARIA CLEAR TYPE",
        ],
        var_name="PATTERN",
        value_name="CLEAR TYPE",
    )
    KEY = ("BEGINNER", "NORMAL", "HYPER", "ANOTHER", "LEGGENDARIA")

    melted_data["PATTERN"] = melted_data["PATTERN"].str.replace(
        " CLEAR TYPE", "", regex=False
    )

    merged_data.append(melted_data.copy(deep=True))
```

df1 내의 같은 곡 다른 패턴을 난이도와 함께 분리


및 병합


```python
TO_MERGE = ("LEVEL", "SCORE", "MISS COUNT")
for idx, df in enumerate((df1, df2)):
    for mg in TO_MERGE:
        melted_data = df.melt(
            id_vars="NAME",
            value_vars=[
                f"BEGINNER {mg}",
                f"NORMAL {mg}",
                f"HYPER {mg}",
                f"ANOTHER {mg}",
                f"LEGGENDARIA {mg}",
            ],
            var_name="PATTERN",
            value_name=mg,
        )

        melted_data["PATTERN"] = melted_data["PATTERN"].str.replace(
            f" {mg}", "", regex=False
        )

        if mg == "LEVEL":
            merged_data[idx] = merged_data[idx].merge(
                right=melted_data[melted_data[mg] != 0],
                on=["NAME", "PATTERN"],
                how="inner",
            )
        else:
            merged_data[idx] = merged_data[idx].merge(
                right=melted_data,
                on=["NAME", "PATTERN"],
                how="inner",
            )
            if mg == "MISS COUNT":
                merged_data[idx]["MISS COUNT"] = merged_data[idx]["MISS COUNT"].replace(
                    "---", -1
                )
                merged_data[idx]["MISS COUNT"] = pd.to_numeric(
                    merged_data[idx]["MISS COUNT"]
                )

merged_data1 = merged_data[0]
merged_data2 = merged_data[1]
```

곡명->패턴순 정렬


```python
KEY = ("BEGINNER", "NORMAL", "HYPER", "ANOTHER", "LEGGENDARIA")

merged_data1.sort_values(
    by=["NAME", "PATTERN"],
    key=lambda s: s.map(lambda x: KEY.index(x)) if s.name == "PATTERN" else s,
)
```

```python
from matplotlib import rcParams

CLEARTYPE_ORDER = (
    "NO PLAY",
    "FAILED",
    "ASSIST CLEAR",
    "EASY CLEAR",
    "CLEAR",
    "HARD CLEAR",
    "EX HARD CLEAR",
    "FULLCOMBO CLEAR",
)
df1_ct = (
    merged_data1.groupby(by="CLEAR TYPE")["PATTERN"].count().reindex(CLEARTYPE_ORDER)
).fillna(0)
df2_ct = (
    merged_data2.groupby(by="CLEAR TYPE")["PATTERN"].count().reindex(CLEARTYPE_ORDER)
).fillna(0)

if isinstance(df1_ct, pd.Series) and isinstance(df2_ct, pd.Series):
    plt.figure(figsize=(1024, 768, "px"))
    rcParams["font.family"] = "Pretendard GOV"

    p1 = plt.bar(
        x=np.arange(len(df1_ct)) - 0.15,
        height=df1_ct,
        label="33 Sparkle Shower",
        width=0.3,
    )
    p2 = plt.bar(
        x=np.arange(len(df2_ct)) + 0.15,
        height=df2_ct,
        label="34 ZINRAI",
        width=0.3,
    )
    plt.xticks(
        np.arange(len(df2_ct)),
        df2_ct.index.to_list(),
    )

    plt.bar_label(p1, label_type="edge")
    plt.bar_label(p2, label_type="edge")
    plt.legend()
    plt.xlabel("CLEAR TYPE")
    plt.ylabel("COUNT")
    plt.show()
```

```python
from matplotlib import rcParams

df1_clear: pd.Series = (
    merged_data1[
        (merged_data1["CLEAR TYPE"] != "NO PLAY")
        & (merged_data1["CLEAR TYPE"] != "FAILED")
    ]
    .groupby(by="LEVEL")["PATTERN"]
    .count()
)
df2_clear: pd.Series = (
    merged_data2[
        (merged_data2["CLEAR TYPE"] != "NO PLAY")
        & (merged_data2["CLEAR TYPE"] != "FAILED")
    ]
    .groupby(by="LEVEL")["PATTERN"]
    .count()
)
df1_failed: pd.Series = (
    merged_data1[merged_data1["CLEAR TYPE"] == "FAILED"]
    .groupby(by="LEVEL")["PATTERN"]
    .count()
)
df2_failed: pd.Series = (
    merged_data2[merged_data2["CLEAR TYPE"] == "FAILED"]
    .groupby(by="LEVEL")["PATTERN"]
    .count()
)

plt.figure(figsize=(1024, 768, "px"))
rcParams["font.family"] = "Pretendard GOV"
SCALE = 1.5

p1 = plt.bar(
    x=np.arange(len(df1_clear)) * SCALE - 0.45,
    height=df1_clear,
    label="33 Sparkle Shower",
    width=0.3,
)
p2 = plt.bar(
    x=np.arange(len(df2_clear)) * SCALE + 0.15,
    height=df2_clear,
    label="34 ZINRAI",
    width=0.3,
)
p3 = plt.bar(
    x=np.arange(len(df1_failed)) * SCALE - 0.15,
    height=df1_failed,
    label="33 Sparkle Shower FAILED",
    width=0.3,
)
p4 = plt.bar(
    x=np.arange(len(df2_failed)) * SCALE + 0.45,
    height=df2_failed,
    label="34 ZINRAI FAILED",
    width=0.3,
)
plt.xticks(
    np.arange(len(df2_clear)) * SCALE,
    df2_clear.index.to_list(),
)

plt.bar_label(p1, label_type="edge")
plt.bar_label(p2, label_type="edge")
plt.bar_label(p3, label_type="edge")
plt.bar_label(p4, label_type="edge")
plt.legend()
plt.xlabel("LEVEL")
plt.ylabel("COUNT")
plt.show()
```
