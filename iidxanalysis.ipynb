## CSV 불러오기


```python
# 여기를 수정
file1 = Path("./data/1.csv")
file2 = Path("./data/2.csv")

df1 = pd.read_csv(file1)
df2 = pd.read_csv(file2)

MAPPER = {
    "バージョン": "VERSION",
    "タイトル": "NAME",
    "ジャンル": "GENRE",
    "アーティスト": "ARTIST",
    "プレー回数": "PLAY COUNT",
    "BEGINNER 難易度": "BEGINNER LEVEL",
    "BEGINNER スコア": "BEGINNER SCORE",
    "BEGINNER ミスカウント": "BEGINNER MISS COUNT",
    "BEGINNER クリアタイプ": "BEGINNER CLEAR TYPE",
    "NORMAL 難易度": "NORMAL LEVEL",
    "NORMAL スコア": "NORMAL SCORE",
    "NORMAL ミスカウント": "NORMAL MISS COUNT",
    "NORMAL クリアタイプ": "NORMAL CLEAR TYPE",
    "HYPER 難易度": "HYPER LEVEL",
    "HYPER スコア": "HYPER SCORE",
    "HYPER ミスカウント": "HYPER MISS COUNT",
    "HYPER クリアタイプ": "HYPER CLEAR TYPE",
    "ANOTHER 難易度": "ANOTHER LEVEL",
    "ANOTHER スコア": "ANOTHER SCORE",
    "ANOTHER ミスカウント": "ANOTHER MISS COUNT",
    "ANOTHER クリアタイプ": "ANOTHER CLEAR TYPE",
    "LEGGENDARIA 難易度": "LEGGENDARIA LEVEL",
    "LEGGENDARIA スコア": "LEGGENDARIA SCORE",
    "LEGGENDARIA ミスカウント": "LEGGENDARIA MISS COUNT",
    "LEGGENDARIA クリアタイプ": "LEGGENDARIA CLEAR TYPE",
    "最終プレー日時": "LAST PLAYED",
}

df1 = df1.rename(columns=MAPPER)
df2 = df2.rename(columns=MAPPER)
```

df1 내의 같은 곡 다른 패턴을 CLEAR TYPE과 함께 분리


```python
melted_data = df1.melt(
    id_vars="NAME",
    value_vars=[
        "BEGINNER CLEAR TYPE",
        "NORMAL CLEAR TYPE",
        "HYPER CLEAR TYPE",
        "ANOTHER CLEAR TYPE",
        "LEGGENDARIA CLEAR TYPE",
    ],
    var_name="PATTERN",
    value_name="CLEAR TYPE",
)
KEY = ("BEGINNER", "NORMAL", "HYPER", "ANOTHER", "LEGGENDARIA")

melted_data["PATTERN"] = melted_data["PATTERN"].str.replace(
    " CLEAR TYPE", "", regex=False
)

merged_data1 = melted_data.copy(deep=True)
```

df1 내의 같은 곡 다른 패턴을 난이도와 함께 분리


및 병합


```python
melted_data = df1.melt(
    id_vars="NAME",
    value_vars=[
        "BEGINNER LEVEL",
        "NORMAL LEVEL",
        "HYPER LEVEL",
        "ANOTHER LEVEL",
        "LEGGENDARIA LEVEL",
    ],
    var_name="PATTERN",
    value_name="LEVEL",
)

melted_data["PATTERN"] = melted_data["PATTERN"].str.replace(" LEVEL", "", regex=False)

merged_data1 = merged_data1.merge(
    right=melted_data[melted_data["LEVEL"] != 0],
    on=["NAME", "PATTERN"],
    how="inner",
)
```

```python
melted_data = df1.melt(
    id_vars="NAME",
    value_vars=[
        "BEGINNER SCORE",
        "NORMAL SCORE",
        "HYPER SCORE",
        "ANOTHER SCORE",
        "LEGGENDARIA SCORE",
    ],
    var_name="PATTERN",
    value_name="SCORE",
)

melted_data["PATTERN"] = melted_data["PATTERN"].str.replace(" SCORE", "", regex=False)

merged_data1 = merged_data1.merge(
    right=melted_data,
    on=["NAME", "PATTERN"],
    how="inner",
)
```

```python
melted_data: pd.DataFrame = df1.melt(
    id_vars="NAME",
    value_vars=[
        "BEGINNER MISS COUNT",
        "NORMAL MISS COUNT",
        "HYPER MISS COUNT",
        "ANOTHER MISS COUNT",
        "LEGGENDARIA MISS COUNT",
    ],
    var_name="PATTERN",
    value_name="MISS COUNT",
)

melted_data["PATTERN"] = melted_data["PATTERN"].str.replace(
    " MISS COUNT", "", regex=False
)

merged_data1 = merged_data1.merge(
    right=melted_data,
    on=["NAME", "PATTERN"],
    how="inner",
)
merged_data1["MISS COUNT"] = merged_data1["MISS COUNT"].replace("---", -1)
merged_data1["MISS COUNT"] = pd.to_numeric(merged_data1["MISS COUNT"])
```
