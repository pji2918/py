## 데이터 전처리


df1, df2를 저장할 리스트 선언


```python
merged_data: list[pd.DataFrame] = []
```

df1 내의 같은 곡 다른 패턴을 CLEAR TYPE과 함께 분리


```python
for df in (df_sps, df_zin):
    melted_data = df.melt(
        id_vars="NAME",
        value_vars=[
            "BEGINNER CLEAR TYPE",
            "NORMAL CLEAR TYPE",
            "HYPER CLEAR TYPE",
            "ANOTHER CLEAR TYPE",
            "LEGGENDARIA CLEAR TYPE",
        ],
        var_name="PATTERN",
        value_name="CLEAR TYPE",
    )
    KEY = ("BEGINNER", "NORMAL", "HYPER", "ANOTHER", "LEGGENDARIA")

    melted_data["PATTERN"] = melted_data["PATTERN"].str.replace(
        " CLEAR TYPE", "", regex=False
    )

    merged_data.append(melted_data.copy(deep=True))
```

df1 내의 같은 곡 다른 패턴을 난이도와 함께 분리  
및 병합


```python
TO_MERGE = ("LEVEL", "DJ LEVEL", "SCORE", "MISS COUNT")
for idx, df in enumerate((df_sps, df_zin)):
    for mg in TO_MERGE:
        melted_data = df.melt(
            id_vars="NAME",
            value_vars=[
                f"BEGINNER {mg}",
                f"NORMAL {mg}",
                f"HYPER {mg}",
                f"ANOTHER {mg}",
                f"LEGGENDARIA {mg}",
            ],
            var_name="PATTERN",
            value_name=mg,
        )

        melted_data["PATTERN"] = melted_data["PATTERN"].str.replace(
            f" {mg}", "", regex=False
        )

        if mg == "LEVEL":
            merged_data[idx] = merged_data[idx].merge(
                right=melted_data[melted_data[mg] != 0],
                on=["NAME", "PATTERN"],
                how="inner",
            )
        else:
            merged_data[idx] = merged_data[idx].merge(
                right=melted_data,
                on=["NAME", "PATTERN"],
                how="inner",
            )
            if mg == "MISS COUNT":
                merged_data[idx][mg] = merged_data[idx][mg].replace("---", -1)
                merged_data[idx][mg] = pd.to_numeric(merged_data[idx][mg])
            elif mg == "DJ LEVEL":
                merged_data[idx][mg] = merged_data[idx][mg].replace("---", "F")

sps_merged_data = merged_data[0]
zin_merged_data = merged_data[1]
```

# 머신/딥러닝


```python
# 예측하려는 게 CLEAR TYPE이니,
# NO PLAY만 결측치가 아닐까 생각해 봤어.
# 점수가 0이거나, MISS COUNT가 -1인 것도 결국 전 버전에서
# 플레이해본 것이니 말이야.
extracted_datas: pd.DataFrame = zin_merged_data[
    (zin_merged_data["CLEAR TYPE"] != "NO PLAY")
]
extracted_datas
```

```python
extracted_datas["CLEAR TYPE"].value_counts()
```

```python
def map_cleartype(clear_type: str):
    if clear_type == "FAILED":
        return 0
    elif clear_type in ("ASSIST CLEAR", "EASY CLEAR", "CLEAR"):
        return 1
    elif clear_type in ("HARD CLEAR", "EX HARD CLEAR"):
        return 2
    elif clear_type == "FULLCOMBO CLEAR":
        return 3
    else:
        return np.nan


extracted_datas["CLEAR TYPE_ORDINAL"] = extracted_datas["CLEAR TYPE"].map(
    map_cleartype, na_action="ignore"
)
extracted_datas.sample(15)
```

```python
extracted_datas["CLEAR TYPE_ORDINAL"].value_counts().reindex((0, 1, 2, 3))
```

```python
KEY = ("BEGINNER", "NORMAL", "HYPER", "ANOTHER", "LEGGENDARIA")

extracted_datas["PATTERN_ORDINAL"] = extracted_datas["PATTERN"].map(func=KEY.index)

extracted_datas.sample(15)
```

```python
import torch

x = torch.tensor(
    extracted_datas[["LEVEL", "PATTERN_ORDINAL"]].to_numpy(), dtype=torch.float32
)
y = torch.tensor(data=extracted_datas["CLEAR TYPE_ORDINAL"].to_numpy())

print(x.shape, x.dtype, y.shape, y.dtype, sep="\n")
```

```python
class IIDXDataSet(torch.utils.data.Dataset):
    def __init__(self, x: torch.Tensor, y: torch.Tensor):
        if len(x) != len(y):
            raise ValueError("Invalid data; x-y doesn't match")
        self.x = x
        self.y = y

    def __getitem__(self, index: int):
        return self.x[index], self.y[index]

    def __len__(self):
        return len(self.x)
```

```python
ds = IIDXDataSet(x, y)
print(ds[0])
```
